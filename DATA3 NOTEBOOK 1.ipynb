```python
import pandas as pd

base_url = "https://footballdata.io/api/v1/players"
page = 1
all_players = []

while True:
    response = requests.get(
        base_url,
        headers=headers,
        params={"page": page, "limit": 100},
        timeout=30,
    )
    response.raise_for_status()
    page_data = response.json()
    all_players.extend(page_data.get("data", []))

    pagination = page_data.get("meta", {}).get("pagination", {})
    total_pages = pagination.get("total_pages", page)
    if page >= total_pages:
        break
    page += 1

newdata_df = pd.json_normalize(all_players)
print(f"Loaded {len(newdata_df)} of {pagination.get('total', len(newdata_df))} players")
display(newdata_df)
```
